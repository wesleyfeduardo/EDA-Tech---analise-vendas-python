```python
import pandas as pd
```

```python
#Exibindo as 5 primeiras linhas
df = pd.read_csv("/content/vendas.csv")
df
```

```python
df.info()
```

```python
df.head()
```

```python
#Número totao de linhas e colunas
df.shape
```

```python
#Calculando receita total
df["receita_total"] = df["quantidade"] * df["preco_unitario"]
df["receita_total"]
```

```python
#Filtrando e exibindo as vendas da categoria Eletrônicos
df_eletronicos = df.loc[df["categoria"] == "Eletrônicos"]
df_eletronicos
```

```python
#Produto mais vendido em quantidade
produto_mais_vendido = df.groupby("produto")[["quantidade"]].sum().sort_values(by="quantidade", ascending=False).head(1).reset_index()
produto_mais_vendido
```

```python
#Identificando a região com maior valor de compra
df_regiao = df.groupby("regiao")[["receita_total"]].sum().sort_values(by="receita_total", ascending=False).head(1).reset_index()
df_regiao
```
